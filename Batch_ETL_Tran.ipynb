```python
import requests
import pandas as pd

from Data_Connection import get_connection
```

```python
API_URL = "https://fakestoreapi.com/products"

response = requests.get(API_URL)

response.raise_for_status()

data = response.json()

print("API Data Extracted Successfully")
```

```python
df = pd.DataFrame(data)

df.head()
```

```python
df = df[
    [
        "id",
        "title",
        "price",
        "category",
        "rating"
    ]
]

df["rating"] = df["rating"].apply(lambda x: x["rate"])

df.columns = [
    "ProductID",
    "ProductName",
    "Price",
    "Category",
    "Rating"
]

df = df.drop_duplicates()

df = df.dropna()

df.head()
```

```python
conn = get_connection()

cursor = conn.cursor()

print("Connected Successfully")
```

```python
cursor.execute("""
CREATE TABLE IF NOT EXISTS fake_store.products
(
    ProductID INT,
    ProductName STRING,
    Price DOUBLE,
    Category STRING,
    Rating DOUBLE
)
""")

insert_query = """
INSERT INTO fake_store.products
VALUES (?, ?, ?, ?, ?)
"""

rows = list(df.itertuples(index=False, name=None))

cursor.executemany(insert_query, rows)

conn.commit()

print("Data Stored Successfully")
```

```python
cursor.execute("""
SELECT *
FROM fake_store.products
LIMIT 10
""")

rows = cursor.fetchall()

for row in rows:
    print(row)
```

```python
cursor.close()
conn.close()
```
